Script Overview
1. Locates the sitemap with all the URLs
2. Download and save HTML content of each URL

```python
import os 
import xmltodict
import requests
```

1. Locate sitemap

```python
# specify the file path 
site_map_file_path = "sitemaps_manual//urls_9_12_2024.xml"

with open(site_map_file_path, 'r') as file:
    sitemap = xmltodict.parse(file.read())
```

2. Download and save HTML content of each URL

```python
"""
This script iterates through a list of URLs extracted from a manually made sitemap  downloads each page's HTML content, 
and saves it as an .html file in the specified output directory. It skips files that have already been downloaded 
to avoid duplicates. URLs that fail to download are collected and printed at the end for review.
SSL verification is disabled for handling certain HTTPS issues.
"""'

urls = sitemap["urls"]["url"]
output_dir = "html_files"
error_URLS = []


for url in urls:
    try:
        # create a valid filename from the URL 
        file_name = url.split("/")[-1] + ".html"
        file_path = os.path.join(output_dir, file_name)
        
        # check it already exists 
        if os.path.exists(file_path):
            print(f"File {file_name} already exists. Skipping Download")
            continue
        
        
        
        page_response = requests.get(url, verify = False)

        if page_response.status_code == 200:
            page_content = page_response.text
            
            # save the HTML content to a file

            with open(file_path, "w", encoding = "utf-8") as file:
                file.write(page_content)

            print(f"Saved HTML for {url} to {file_path}")
        
        else:
            print(f"Failed to retreive {url}. Status code: {page_response.status_code}")
            error_URLS.append(urls)
    
    except Exception as e:
        print(f"Error downloading {url}: {e}")
        error_URLS.append(url)
        continue


if error_URLS:
    print("\nError URLS:")
    for url in error_URLS:
        print(url)

```
