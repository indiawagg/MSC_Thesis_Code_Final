Final Script below 

```python

"""
This script scrapes article URLs from the Australian Bureau of Statistics (ABS) articles page
using Selenium for automatic page handling. It collects the URLs of all articles listed across
pages and saves them into an XML file called 'urls.xml' The URL was then renamed to add the date of the collection of the URLs.
"""


from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
import xml.etree.ElementTree as ET

# Set up the Chrome driver path and initialize the driver
chrome_driver_path = r"C:\Users\india\OneDrive\Documents\Thesis MSC\Python scraping\chromedriver-win64\chromedriver-win64\chromedriver.exe"
service = Service(executable_path=chrome_driver_path)
driver = webdriver.Chrome(service=service)

# Go to the URL
base_url = "https://www.abs.gov.au"
driver.get("https://www.abs.gov.au/articles")

# Let the page load
time.sleep(3)

# Create the root XML element
root = ET.Element("urls")

# Loop through pages and collect URLs
hrefs = []
try:
    while True:
        # Get all the links on the current page
        content_div = driver.find_element(By.ID, "content")
        links = content_div.find_elements(By.TAG_NAME, "h2")

        for link in links:
            a_tag = link.find_element(By.TAG_NAME, "a")
            if a_tag:
                href = a_tag.get_attribute("href")
                if href:
                    
                    full_url = href if href.startswith("https") else base_url + href

                    # Create XML element for each URL
                    url_element = ET.SubElement(root, "url")
                    url_element.text = full_url
                    hrefs.append(full_url)

        # Check if there is a "Next" button to go to the next page
        next_button = driver.find_element(By.XPATH, '//a[@rel="next"]')
        if next_button:
            next_button.click()
            print("Navigated to the next page.")
            time.sleep(3)  # Wait for the page to load
        else:
            print("No more pages found.")
            break

except Exception as e:
    print(f"Error occurred: {e}")

# Create the XML tree and save it to a file
file_path = "urls.xml"
tree = ET.ElementTree(root)

# Write the XML data to the file
with open(file_path, "wb") as xml_file:
    tree.write(xml_file, encoding="utf-8", xml_declaration=True)

# Print the list of URLs for verification
print(hrefs)

# Close the driver
driver.quit()
```
