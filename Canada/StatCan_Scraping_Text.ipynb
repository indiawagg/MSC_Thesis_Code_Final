Script Overview:
- The script scrapes the text from the previously downloaded HTML content

```python
"""
This script extracts and saves structured text from local HTML files, originally downloaded 
from web articles. It parses each HTML file to retrieve:
- The date the article was published
- The title of the page
- The main article content 

If an error occurs during processing, the filename is kept.
"""


import os
from bs4 import BeautifulSoup

html_dir = "html_files"
output_dir = "scraped_articles"
error_files = []

html_files = [f for f in os.listdir(html_dir) if f.endswith(".html")]


for html_file in html_files:
    file_path = os.path.join(html_dir, html_file)
    output_file_path = os.path.join(output_dir, f"{os.path.splitext(html_file)[0]}.txt")

    if os.path.exists(output_file_path):
        print(f"{output_file_path} already exists. Skipping this file.")
        continue

    try:
       
        with open(file_path, "r", encoding="utf-8") as f:
            page_content = f.read()

        soup = BeautifulSoup(page_content, "html.parser")

        date = soup.find(class_="sd-release-date")
        if date:
            date = date.get_text().replace("\xa0", "").replace("Released:", "")
        else:
            date = "Date not found"
        
        header = soup.find("head")
        if header:
            title = header.find("title")
            if title:
                title = title.get_text(strip=True)
            else:
                title = "Title not found"
        else:
            title = "Header not found"        

        
        main = soup.find("main")
        if main:
            sections = main.find_all("section")
            if len(sections) > 1:
                content = sections[1]  

               
                for div in content.find_all('div'):
                    div.decompose()  

          
                text_sections = content.find_all("p")
                text_content = "\n".join([p.get_text() for p in text_sections])
            else:
                text_content = "Article content not found"
        else:
            text_content = "Main section not found"

        with open(output_file_path, "w", encoding = "utf-8") as out_file:
            out_file.write(f"Date:\n{date}\n")
            out_file.write(f"Article Title:\n{title}\n")
            out_file.write(f"Article Content:\n{text_content}\n")

        print(f"File has been saved at {output_file_path}")
         

    except Exception as e:
        print(f"Error processing {html_file}: {e}")
```
