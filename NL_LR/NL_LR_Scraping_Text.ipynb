Final Chromedriver Script : Used this one


```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
from selenium.common.exceptions import NoSuchElementException

# Chromedriver Path
chrome_driver_path = r"C:\Users\india\OneDrive\Documents\Thesis MSC\Python scraping\chromedriver-win64\chromedriver-win64\chromedriver.exe"
service = Service(executable_path= chrome_driver_path)
driver = webdriver.Chrome(service = service)

#initialize starting URL 

URL = "https://longreads.cbs.nl/im2018-1/"
driver.get(URL)

# Set a storage for all content 
all_content = ""

# function text scraper
def text_scraper():
    # raw HTML string
    page_source = driver.page_source
    
    soup = BeautifulSoup(page_source, "html.parser")
    main = soup.find("article")

    if main:

        for a in main.find_all("a"):
            a.decompose()

        for script in main.find_all('script'):
            script.decompose() 

        text_sections = main.find_all(["p","li","h1","h2","h3"])
        text = "\n\n".join([text_section.get_text(separator = " ",strip = True) for text_section in text_sections])
        text = "\n\n".join([" ".join(paragraph.split()) for paragraph in text.split("\n\n")])

        return text
    
    else:
        return ""
    
all_content += text_scraper()

starting_url = driver.current_url

# Pagination

while True:

    try:
        next_page_link = driver.find_element(By.CLASS_NAME , "pagination-next")

        # checking the current url is not the starting one and that content is not empty
        if "disabled" in next_page_link.get_attribute("class"):
            print("No more pages available")
            break

        next_page_link.click()
        time.sleep(2)

        if driver.current_url == starting_url and all_content != "":
            print(f"Reached the starting URL again: {starting_url}")
            break

        all_content += text_scraper()
    
    except NoSuchElementException:
        print("No more pages available")
        break


file_path= r"C:\Users\india\OneDrive\Documents\Thesis MSC\Python scraping\articles\De_positie_van_Nederland"

with open(file_path, "w", encoding = "utf-8") as file:
    file.write(f"Article Content:\n {all_content}\n")
    print(f"All pages have been scraped and saved at {file_path}")

driver.quit()
```
